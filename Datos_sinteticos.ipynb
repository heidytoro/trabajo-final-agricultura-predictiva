4. Función para generar variables químicas

```python
import numpy as np # Ensure numpy is imported if not already.

def generar_quimica(estado):
    # Initialize ce, clase_p, clase_k based on estado
    if estado == "Bueno":
        ce = np.random.uniform(0.1, 1.5)
        clase_p = np.random.choice(["Optimo", "Alto"], p=[0.40, 0.60])
        clase_k = np.random.choice(["Optimo", "Alto"], p=[0.45, 0.55])
    elif estado == "Regular":
        ce = np.random.uniform(0.05, 2.0)
        clase_p = np.random.choice(["Bajo", "Optimo", "Alto"], p=[0.40, 0.35, 0.25])
        clase_k = np.random.choice(["Bajo", "Optimo", "Alto"], p=[0.35, 0.45, 0.20])
    else: # estado == "Deficiente"
        ce = np.random.uniform(0.01, 3.0)
        clase_p = np.random.choice(["Bajo", "Optimo"], p=[0.75, 0.25])
        clase_k = np.random.choice(["Bajo", "Optimo"], p=[0.80, 0.20])

    # Now assign p, ph, mo, cic based on clase_p (from the new code provided by the user)
    if clase_p == "Bajo":
        p = np.random.uniform(3, 18)
        ph = np.random.uniform(4.2, 5.4)
        mo = np.random.uniform(0.5, 2.5)
        cic = np.random.uniform(1, 9)
    elif clase_p == "Optimo":
        p = np.random.uniform(22, 38)
        ph = np.random.uniform(5.5, 6.5)
        mo = np.random.uniform(2.8, 5.0)
        cic = np.random.uniform(8, 18)
    else: # clase_p == "Alto"
        p = np.random.uniform(42, 60)
        ph = np.random.uniform(6.0, 7.0)
        mo = np.random.uniform(4.0, 6.5)
        cic = np.random.uniform(15, 28)

    # Assign k based on clase_k (retained from original structure)
    if clase_k == "Bajo":
        k = np.random.uniform(0.05, 0.27)
    elif clase_k == "Optimo":
        k = np.random.uniform(0.35, 0.95)
    else: # clase_k == "Alto"
        k = np.random.uniform(1.05, 1.50)

    return [round(x, 2) for x in [ph, mo, p, k, cic, ce]]
```

5. Generar la base de datos

```python
# 50% de los registros serán "Bueno".
# 35% de los registros serán "Regular".
# 15% de los registros serán "Deficiente".

estados = ["Bueno", "Regular", "Deficiente"]
pesos = [0.45, 0.40, 0.15]

datos = []# Lista vacía donde se almacenarán todas las filas de la base de datos

for _ in range(n):
    municipio = random.choice(municipios) # Selecciona aleatoriamente un municipio de la lista "municipios".
    municipio_cod = municipios_map[municipio]# Busca el código numérico asociado al municipio seleccionado.
    departamento = random.choice(departamentos)# Selecciona aleatoriamente un departamento de la lista "departamentos".
    departamento_cod = departamentos_cod[departamento.lower()] # Obtiene el código numérico del departamento.
    estado = random.choices(estados, weights=pesos, k=1)[0]# Selecciona el estado del cultivo usando las probabilidades definidas.

    tiempo_txt, tiempo_cod = random.choice(list(tiempos.items()))# Selecciona aleatoriamente una categoría de tiempo de establecimiento
    topo_cod = random.randint(0, 3) # 0 = Plana, 1 = Ondulada, 2 = Quebrada, 3 = Montañosa
    dren_cod = random.randint(0, 3)# 0 = Deficiente, 1 = Moderado, 2 = Bueno, 3 = Excelente
    riego_cod = random.randint(0, 3)#   0 = No, 1 = Ocasional, 2 = Frecuente, 3 = Tecnificado

    quimica = generar_quimica(estado)

    # Generate 'aplica_' values randomly (0 or 1)
    aplica_p = np.random.randint(0, 2)
    aplica_k = np.random.randint(0, 2)
    aplica_cal = np.random.randint(0, 2)
    aplica_organico = np.random.randint(0, 2)
    aplica_kcl = np.random.randint(0, 2)
    fert_generica = np.random.randint(0, 2)

    fila = [
        departamento_cod,
        municipio,
        municipio_cod,
        "Caña panelera",
        estado,
        tiempo_txt,
        tiempo_cod,
        topo_cod,
        dren_cod,
        riego_cod,
        f"{aplica_p}, {aplica_k}, {aplica_cal}, {aplica_organico}, {aplica_kcl}", # Removed fert_generica from this string
        *quimica, # Unpack the 6 chemical values here: pH, MO, P, K, CIC, CE
        aplica_p,
        aplica_k,
        aplica_cal,
        aplica_organico,
        aplica_kcl,
        fert_generica # Added fert_generica as a separate element
    ]

    datos.append(fila)
```

```python
import pandas as pd # Ensure pandas is imported for pd.DataFrame.head() if not already done

# Diccionarios de referencia, similares a la codificacion usada en tiempo.
fosforo_category_map = {
    "Bajo": 0,      # < 20 mg/kg
    "Optimo": 1,   # 20 - 40 mg/kg
    "Alto": 2       # > 40 mg/kg
}


def clasificar_fosforo(valor):
    if valor < 20:
        return fosforo_category_map["Bajo"]
    elif valor <= 40:
        return fosforo_category_map["Optimo"]
    else:
        return fosforo_category_map["Alto"]


# Crear columnas categoricas codificadas en todo el dataset.
df["fosforo_category"] = df["Fosforo Bray II"].apply(clasificar_fosforo)


# Crear columnas con etiquetas de texto tipo category para usarlas como variable objetivo.
# Esto permite usar: y_test_fosforo_category.cat.categories.tolist()
df["fosforo_category_label"] = df["fosforo_category"].map({
    0: "Bajo",
    1: "Óptimo",
    2: "Alto"
}).astype("category")


print("Distribucion de fosforo_category:")
display(df["fosforo_category"].value_counts().sort_index().rename_axis("Clase").reset_index(name="Cantidad"))



display(df[["Fosforo Bray II", "fosforo_category", "fosforo_category_label"]].head(10))
```

##Optimización de Hiperparámetros para Fósforo (GridSearchCV)

```python
from sklearn.model_selection import GridSearchCV
from sklearn.ensemble import RandomForestClassifier
import pandas as pd # Import pandas for DataFrame creation

print("Realizando la búsqueda de hiperparámetros con GridSearchCV para 'fosforo_category'...")

# Definir el modelo base
rf_model_grid = RandomForestClassifier(random_state=42, n_jobs=-1)

# Definir la cuadrícula de hiperparámetros a buscar
param_grid = {
    'n_estimators': [100, 200, 300], # Number of trees in the forest
    'max_depth': [10, 15, 20],      # Maximum depth of the tree
    'min_samples_split': [2, 5, 10], # Minimum number of samples required to split an internal node
    'min_samples_leaf': [1, 2, 4]    # Minimum number of samples required to be at a leaf node
}

# Configurar GridSearchCV
grid_search_fosforo = GridSearchCV(
    estimator=rf_model_grid,
    param_grid=param_grid,
    cv=3, # 3-fold cross-validation
    scoring='accuracy', # Use accuracy as the scoring metric
    n_jobs=-1, # Use all available cores
    verbose=2
)

# Ejecutar la búsqueda en el conjunto de entrenamiento escalado
grid_search_fosforo.fit(X_train_fosforo_scaled, y_train_fosforo_category)

print("\nBúsqueda de hiperparámetros completada.")

# Mostrar los mejores parámetros y la mejor puntuación
print(f"Mejores hiperparámetros encontrados: {grid_search_fosforo.best_params_}")
print(f"Mejor puntuación (accuracy) en validación cruzada: {grid_search_fosforo.best_score_:.4f}")

# Obtener el mejor modelo
best_rf_model_fosforo = grid_search_fosforo.best_estimator_

# Evaluar el mejor modelo en el conjunto de prueba
y_pred_best_fosforo = best_rf_model_fosforo.predict(X_test_fosforo_scaled)

# Imprimir el classification report del mejor modelo
from sklearn.metrics import classification_report
print("\nClassification Report del mejor modelo RandomForestClassifier para 'fosforo_category':\n")
print(classification_report(y_test_fosforo_category, y_pred_best_fosforo))

# Obtener las importancias de las características del mejor modelo
feature_importances = best_rf_model_fosforo.feature_importances_

# Obtener los nombres de las características
feature_names = X_train_fosforo_scaled.columns

# Crear un DataFrame para visualizar las importancias
importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': feature_importances
})

# Ordenar las características por importancia de forma descendente
importance_df = importance_df.sort_values(by='Importance', ascending=False)

print("\nVariables más importantes para la predicción de 'fosforo_category':")
display(importance_df)

```

## Modelo de Clasificación para Fósforo (SVC)

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler
from sklearn.svm import SVC
from sklearn.impute import SimpleImputer
from sklearn.metrics import confusion_matrix, classification_report, accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns

print("Creando y entrenando el modelo SVC para 'fosforo_category'...")

# Definir el pipeline del modelo SVC
modelo_svc_fosforo = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', RobustScaler()),
    ('svc', SVC(
        kernel='rbf',
        C=10,
        gamma='scale',
        probability=True,
        class_weight='balanced',
        random_state=42
    ))
])

# Entrenar el modelo con los datos escalados de fósforo
modelo_svc_fosforo.fit(X_train_fosforo_scaled, y_train_fosforo_category)

print("Modelo SVC entrenado.")

# Realizar predicciones en el conjunto de prueba escalado
y_pred_svc_fosforo = modelo_svc_fosforo.predict(X_test_fosforo_scaled)

# Evaluar el rendimiento del modelo
accuracy_svc_fosforo = accuracy_score(y_test_fosforo_category, y_pred_svc_fosforo)

# Obtener las etiquetas de las clases en el orden deseado
class_labels_fosforo = ["Bajo", "Óptimo", "Alto"]

report_svc_fosforo = classification_report(
    y_test_fosforo_category,
    y_pred_svc_fosforo,
    labels=class_labels_fosforo, # Usar etiquetas de cadena
    target_names=class_labels_fosforo,
    zero_division=0
)

print(f"\nAccuracy para 'fosforo_category' (SVC): {accuracy_svc_fosforo:.4f}")
print("\nClassification Report para 'fosforo_category' (SVC):\n")
print(report_svc_fosforo)

print("\nGenerando la matriz de confusión para 'fosforo_category' (SVC)...")

# Calcular la matriz de confusión
cm_svc_fosforo = confusion_matrix(
    y_test_fosforo_category,
    y_pred_svc_fosforo,
    labels=class_labels_fosforo # Usar etiquetas de cadena
)

# Crear un mapa de calor
plt.figure(figsize=(8, 6))
sns.heatmap(
    cm_svc_fosforo,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_labels_fosforo,
    yticklabels=class_labels_fosforo
)

plt.title('Matriz de Confusión para Fosforo Category (SVC)')
plt.xlabel('Predicción')
plt.ylabel('Valor Verdadero')
plt.show()

print("Matriz de confusión para 'fosforo_category' (SVC) generada exitosamente.")
```

## Modelo de Clasificación para Fósforo (XGBoostClassifier)

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

print("Creando y entrenando el modelo XGBoostClassifier para 'fosforo_category'...")

# Identificar columnas categóricas
categorical_cols = X_train_fosforo_scaled.select_dtypes(include=['category', 'object']).columns

# Convertir columnas categóricas a tipo entero para XGBoost
X_train_fosforo_scaled_xgb = X_train_fosforo_scaled.copy()
X_test_fosforo_scaled_xgb = X_test_fosforo_scaled.copy()

for col in categorical_cols:
    if str(X_train_fosforo_scaled_xgb[col].dtype) == 'category':
        X_train_fosforo_scaled_xgb[col] = X_train_fosforo_scaled_xgb[col].cat.codes.astype(int)
        X_test_fosforo_scaled_xgb[col] = X_test_fosforo_scaled_xgb[col].cat.codes.astype(int)
    else:
        categorias = X_train_fosforo_scaled_xgb[col].astype('category').cat.categories
        X_train_fosforo_scaled_xgb[col] = X_train_fosforo_scaled_xgb[col].astype('category').cat.codes.astype(int)
        X_test_fosforo_scaled_xgb[col] = pd.Categorical(
            X_test_fosforo_scaled_xgb[col],
            categories=categorias
        ).codes.astype(int)

# Crear el modelo XGBoost
modelo_xgb_fosforo = XGBClassifier(
    objective='multi:softmax',
    num_class=3,
    n_estimators=300,
    learning_rate=0.1,
    max_depth=5,
    eval_metric='mlogloss',
    random_state=42,
    n_jobs=-1
)

# Entrenar el modelo con los datos transformados
modelo_xgb_fosforo.fit(X_train_fosforo_scaled_xgb, y_train_fosforo_category.cat.codes)

print("Modelo XGBoostClassifier entrenado.")

# Realizar predicciones en el conjunto de prueba
y_pred_xgb_fosforo = modelo_xgb_fosforo.predict(X_test_fosforo_scaled_xgb)

# Evaluar el rendimiento del modelo
accuracy_xgb_fosforo = accuracy_score(y_test_fosforo_category.cat.codes, y_pred_xgb_fosforo)

report_xgb_fosforo = classification_report(
    y_test_fosforo_category.cat.codes,
    y_pred_xgb_fosforo,
    labels=[0, 1, 2],
    target_names=y_train_fosforo_category.cat.categories.tolist(), # Use actual category names for labels
    zero_division=0
)

print(f"\nAccuracy para 'fosforo_category' (XGBoost): {accuracy_xgb_fosforo:.4f}")
print("\nClassification Report para 'fosforo_category' (XGBoost):\n")
print(report_xgb_fosforo)

print("\nGenerando la matriz de confusión para 'fosforo_category' (XGBoost)...")

# Calcular la matriz de confusión
cm_xgb_fosforo = confusion_matrix(
    y_test_fosforo_category.cat.codes,
    y_pred_xgb_fosforo,
    labels=[0, 1, 2]
)

# Obtener las etiquetas de las clases del `y_train` para asegurar el orden
class_labels_fosforo = y_train_fosforo_category.cat.categories.tolist()

# Crear heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(
    cm_xgb_fosforo,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=class_labels_fosforo,
    yticklabels=class_labels_fosforo
)

plt.title('Matriz de Confusión para Fosforo Category (XGBoost)')
plt.xlabel('Predicción')
plt.ylabel('Valor Verdadero')
plt.show()

print("Matriz de confusión para 'fosforo_category' (XGBoost) generada exitosamente.")
```
